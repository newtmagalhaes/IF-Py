```python
import pandas as pd

INPUT_FILE = 'downloads/microdados_enem_2024/DADOS/RESULTADOS_2024_CE.csv'

df = pd.read_csv(INPUT_FILE).dropna()

print(f'Total de linhas: {len(df)}')
print(f'Colunas: {df.columns}')
df.head()
```

```python
df.head().transpose()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style('ticks')

COLUNAS_REDACAO = [
    'NU_NOTA_COMP1',
    'NU_NOTA_COMP2',
    'NU_NOTA_COMP3',
    'NU_NOTA_COMP4',
    'NU_NOTA_COMP5',
    'NU_NOTA_REDACAO'
]
```

```python
df[COLUNAS_REDACAO[:-1]].boxplot(figsize=(16, 8))
plt.show()

df[COLUNAS_REDACAO[-1]].plot.box()
plt.show()

df[COLUNAS_REDACAO].describe()
```

```python
from geobr import read_municipality

geodf = read_municipality('CE', year=2024, simplified=True)
geodf.head()
```

```python
dfg = df.groupby('NO_MUNICIPIO_ESC')
dfg[COLUNAS_REDACAO].mean()
```

```python
from IPython.display import display_html

dfg_temp = pd.DataFrame(index=list(dfg.groups.keys()))
dfg_temp['MEAN_REDACAO'] = dfg[['NU_NOTA_REDACAO']].mean()
dfg_temp['STD_REDACAO'] = dfg[['NU_NOTA_REDACAO']].std()
dfg_temp['COUNT_ALUNOS'] = dfg[['NU_NOTA_REDACAO']].count()

display_html(
    dfg_temp
        .sort_values(by='MEAN_REDACAO', ascending=False)
        .style
        .background_gradient(subset=['MEAN_REDACAO'], cmap='OrRd', vmin=0, vmax=1000, axis='columns')
        .to_html(),
    raw=True
)
```

```python
geo_means = geodf.set_index('name_muni').join(dfg_temp)
```

```python
MAIS_DE_100_ALUNOS = dfg_temp['COUNT_ALUNOS'] > 100
notas = [600, 800]
for n in notas:
    media_maior_que_n = dfg_temp['MEAN_REDACAO'] >= n
    filtros = MAIS_DE_100_ALUNOS & media_maior_que_n

    print(f'Média maior que {n}')
    geodf.set_index('name_muni').join(dfg_temp[filtros]).plot(
        column='MEAN_REDACAO',
        legend=True,
        cmap='OrRd',
        vmin=0,
        vmax=1000,
        figsize=(16, 10),
        missing_kwds={
            "color": "lightgrey",
            "edgecolor": "red",
            "hatch": "///",
            "label": "Missing values",
        })
    plt.show()
```
